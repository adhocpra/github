Given a sorted array and a target value, find two numbers that add up to the target.

```python
def two_sum(array,target):
  left=0
  right=len(array)-1
  while left<right:
    current_sum= array[left]+array[right]

    if current_sum==target:
      return [left,right]
    elif current_sum<target:
      left+=1
    elif current_sum>target:
      right-=1
  return[]
print(two_sum([1,2,3,4,5],7))
print(two_sum([1,2,3,4,5],10))
```

Given a sorted array, remove duplicates in-place so each element appears only once.

```python
def remove_duplicate(array):
  if not array:
    return 0
  i=0
  for j in range (1, len(array)):
    if array[j] != array[i]:
      i+=1
      array[i]=array[j]
  return i+1
nums= [1,1,2,2,3,3,4,4]
l1= remove_duplicate(nums)
print(nums[:l1])

```

Problem Statement:
Reverse the elements of a string/array in-place.

```python
def reverse_string(s):
  left=0
  right=len(s)-1
  while left<right:
    s[left],s[right]= s[right],s[left]
    left+=1
    right-=1
  return s
s= ["h","e","l","l","o"]
reverse_string(s)
print(s)
```

Problem Statement:
Move all zeros to the end while maintaining the relative order of non-zero elements.

```python
def move_zeroes(array):
  i=0 #position to write non zero elements
  for j in range (len(array)):
    if array[j] !=0:
      array[i]=array[j]
      i+=1
  while i<len(array):
    array[i]=0
    i+=1
  return array
array=[0,1,0,3,12]
print(move_zeroes(array))


```

Given a sorted array with negative and positive numbers, return array of squares also sorted.

```python
def sorted_square(array):
  result= [0]* len(array)
  left=0
  right= len(array)-1
  for i in range (len(array)-1,-1,-1):
    if abs(array[left])>abs(array[right]):
      result[i]=array[left]**2
      left+=1
    else:
        result[i]=array[right]**2
        right-=1
  return result
print(sorted_square([-4,-1,0,3,10]))
```

```python
#container with most water
def max_area(height):
  left=0
  right= len(height)-1
  max_water=0
  while left<right:
    width= right-left
    area= width*min(height[left], height[right])
    max_water= max(max_water,area)
    if height[left]<height[right]:
      left+=1
    else:
      right-=1
  return max_water
print(max_area([1,8,6,2,5,4,8,3,7]))
print(max_area([1, 1]))                           # Output: 1
print(max_area([4, 3, 2, 1, 4]))
```

Given a string, determine if it is a palindrome (reads the same forwards and backwards), considering only alphanumeric characters and ignoring cases.

```python
def is_palindrome(s):
  cleaned=""
  for char in s:
    if char.isalnum():
      cleaned += char.lower()
  left=0
  right=len(cleaned)-1
  while left<right:
    if cleaned[left] != cleaned[right]:
      return False
    left+=1
    right-=1
  return True
print(is_palindrome("A man, a plan, a canal: Panama"))
print(is_palindrome("race a car"))
```
