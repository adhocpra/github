```python
def decorator(func):
  def wrapper():
    print("before function")
    func()
  return wrapper
@decorator
def greet():
  print("hello")
print(greet.__name__) #this returns wrapper not greet.

```

```python
from functools import wraps
def decorator(func):
  @wraps(func) #using this you will now see name of the function
  def wrapper():
    print("before function")
    func()
  return wrapper
@decorator
def greet():
  print("helllo")
print(greet.__name__)
```

```python
def fibonacci(n): #this is slow because of recursion
  if n<2:
    return n
  return fibonacci(n-1)+fibonacci(n-2)
print(fibonacci(128))
```

```python
from functools import lru_cache #fast
@lru_cache(maxsize=100)
def fibonacci(n):
  if n<2:
    return n
  return fibonacci(n-1)+fibonacci(n-2)
print(fibonacci(128))
```

```python
from itertools import groupby

data= ['a', 'a','a','b','b','b','c','c','a','c','b']
data.sort()
for key,group in groupby(data):
  print(key,list(group))
```

Repeat:

```python
from itertools import repeat
data= repeat("apple",3)
for item in data:
  print(item)
print(list(data))

data2= repeat(2,5)
print(list(data2))

data3= repeat(True,5)
print(next(data3))
print(next(data3))
print(next(data3))
print(next(data3))
print(next(data3))
```

# Collection Module

```python
#Counter:
from collections import Counter
#.most_common:

fruits= ["apple","mango", "banana","apple","banana","banana"]
result= Counter(fruits)
print(result)
print(result["apple"])
print(result["mango"])
print(result["banana"])
print(result.most_common(1))
print(result.most_common(2))
print(result.most_common(3))
```

```python
data= {
    "Python" :["Flask", "Django"] #adding to a dictionary traditional
}
#FastAPI
data["Python"].append("FastAPI")
print(data)
```

```python
data= {}
data["Python"].append(" FastAPI")
print(data)
#Error because of missing key, to handle this use defaultdict
```

```python
from collections import defaultdict
data= defaultdict(list)
data["Python"].append( "FastAPI")
print(data)
data["Python"].append( "Django")
data["Python"].append( "Flask")
print(data)
data["Python"].extend(["NumPy", "Pandas", "Matplotlib"])
print(data)
```

```python
from collections import defaultdict #set
data= defaultdict(set)
data["Skills"]. add("Python")
print(data)
data["Skills"].update(["Java","CSS" ])
print(data)
```

```python
#int
from collections import defaultdict
data= defaultdict(int)
data["apple"] +=3
data["banana"]+=4
#cannot add/change data in normal dict
print(data)
```

```python
from collections import defaultdict

students=[
    ("Python", "A"),
    ("Java", "B"),
    ("Go", "C"),
    ("Python", "D"),
    ("Java", "E"),
    ("Go", "F"),
    ("Python", "G"),
    ("Java", "H"),
    ("Go", "I")
]
data=defaultdict(list)
for language, grade in students:
  data[language].append(grade)
print(data)
print(dict(data))
```

**Deque**

```python
from collections import deque
data= deque( [1,2,3,4,5])
data.pop()
print(data)
data.popleft()
print(data)
data.append(112)
print(data)
data.appendleft(456)
print(data)
```
